```python
#Ask for a year. Print whether it is a leap year. (Rule: divisible by 4, but not by 100, unless also divisible by 400. Example: 2000 and 2024 are leap years, 1900 is not.)
year = int(input("enter your year : "))
if year%4==0 and year%100!=0 or year%400==0:
  print("leap year")
else :
  print("not a leap year")
```
