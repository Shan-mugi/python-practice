```python
def num(a,b,c):
  if a>b and a>c:
    return f" {a} is the largest number"
  elif b>c:
    return f"{b} is the largest number"
  else:
    return f"{c} is the largest number"
ans=num(4,9,2)
print(ans)
```

```python
def calculate_total(name,mark1,mark2,mark3):
  if 0<=mark1<=100 and 0<=mark2<=100 and 0<=mark3<=100:
    total=mark1+mark2+mark3
    avg=total/3
    dic={'name':name,'total':total,'average':avg}
    return dic
  else:
    return "invalid marks"
ans=calculate_total('shan',110,100,100)
print(ans)
```
