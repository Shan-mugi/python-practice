```python
def student(name,mark1,mark2,mark3):
  total=mark1+mark2+mark3
  avg=total/3
  if avg>90:
    print("grade a")
  elif avg>70:
    print("grade b")
  elif avg>45:
    print("grade c")
  else:
    print("fail")
  dic={'name':name,'total':total,'avg':avg}
  return dic
student('shan',56,87,98)



```
